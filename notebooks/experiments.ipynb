```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score,accuracy_score,precision_score,f1_score,recall_score
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import(
    AdaBoostClassifier,
    GradientBoostingClassifier,RandomForestClassifier
)

from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
def evaluate(true,pred):
    accuracy = accuracy_score(true,pred)
    precision = precision_score(true,pred)
    f1 = f1_score(true,pred)
    recall = recall_score(true,pred)

    return accuracy,precision,f1,recall
```

```python
models = {
    "Random Forest" : RandomForestClassifier(verbose=1),
    "Decision Tree" : DecisionTreeClassifier(),
    "Gradient Boosting" : GradientBoostingClassifier(verbose=1),
    "Logistic Regression" : LogisticRegression(verbose=1),
    "AdaBoost" : AdaBoostClassifier(),
    "ANN": make_pipeline(
        StandardScaler(),
        MLPClassifier(
            hidden_layer_sizes=(64, 32),
            activation="relu",
            solver="adam",
            max_iter=300,
            random_state=42,
        ),
    ),
}

model_list = []
accuracy_list = []

for i  in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train,y_train)

    ## Make Predictions
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)

    ## Evaluate

    model_train_accuracy, model_train_precision, model_train_f1,model_train_recall = evaluate(y_train,y_pred_train)
    model_test_accuracy, model_test_precision, model_test_f1,model_test_recall = evaluate(y_test,y_pred_test)

    print(list(models.keys())[i])
    model_list.append(list(models.keys())[i])

    print("---MODEL PERFORMANCE ON TRAINING DATA---")
    print(f"Accuracy {model_train_accuracy:.4f}")
    print(f"Precision {model_train_precision:.4f}")
    print(f"F1 SCORE {model_train_f1:.4f}")
    print(f"Recall {model_train_recall:.4f}")

    print("-------------------------------------")

    print("---MODEL PERFORMANCE ON TEST DATA---")
    print(f"Accuracy {model_test_accuracy:.4f}")
    print(f"Precision {model_test_precision:.4f}")
    print(f"F1 SCORE {model_test_f1:.4f}")
    print(f"Recall {model_test_recall:.4f}")

    accuracy_list.append(model_test_accuracy)

    print('='*35)
    print('\n')
```

```python
models = {
    "Random Forest": RandomForestClassifier(random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "AdaBoost": AdaBoostClassifier(random_state=42),
    "ANN": make_pipeline(
        StandardScaler(),
        MLPClassifier(
            early_stopping=True,
            random_state=42,
        ),
    ),
}



params = {
        "Random Forest": {
            "n_estimators": [100, 200, 300],
            "max_depth": [None, 10, 20],
            "min_samples_split": [2, 5, 10],
            "min_samples_leaf": [1, 2, 4],
            "max_features": ["sqrt", "log2"],
        },

        "Decision Tree": {
            "criterion": ["gini", "entropy"],
            "max_depth": [None, 10, 20, 30],
            "min_samples_split": [2, 5, 10],
            "min_samples_leaf": [1, 2, 4],
        },

        "Gradient Boosting": {
            "n_estimators": [100, 200],
            "learning_rate": [0.01, 0.05, 0.1],
            "max_depth": [2, 3, 5],
            "subsample": [0.8, 1.0],
            "min_samples_split": [2, 5],
        },

        "Logistic Regression": {
            
        },

        "AdaBoost": {
            "n_estimators": [50, 100, 200],
            "learning_rate": [0.01, 0.1, 1.0],
        },

        "ANN": {
        "mlpclassifier__hidden_layer_sizes": [(32,), (64,), (64, 32)],
        "mlpclassifier__activation": ["relu", "tanh"],
        "mlpclassifier__alpha": [0.0001, 0.001],
        "mlpclassifier__learning_rate_init": [0.001, 0.01],
        "mlpclassifier__max_iter": [300],
    },
    }
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score

def evaluate_models(X_train, y_train, X_test, y_test, models, params):
    report = {}

    for name, model in models.items():
        gs = GridSearchCV(
            estimator=model,
            param_grid=params[name],
            cv=3,
            scoring="accuracy",
            n_jobs=-1,
        )

        gs.fit(X_train, y_train)

        best_model = gs.best_estimator_
        y_train_pred = best_model.predict(X_train)
        y_test_pred = best_model.predict(X_test)

        train_accuracy = accuracy_score(y_train, y_train_pred)
        test_accuracy = accuracy_score(y_test, y_test_pred)

        report[name] = test_accuracy

        print(f"{name}:")
        print(f"  Train Accuracy: {train_accuracy:.4f}")
        print(f"  Test Accuracy:  {test_accuracy:.4f}")
        print(f"  Best Parameters: {gs.best_params_}")

    return report
```

```python
report = evaluate_models(
    X_train,
    y_train,
    X_test,
    y_test,
    models,
    params,
)

print(report)
```

```python
best_model_name = max(report, key=report.get)

print(f"Best model: {best_model_name}")
print(f"Test accuracy: {report[best_model_name]:.4f}")
```
